```python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
```

```python
img = Image.open('som.png').convert('L')
image_matrix = np.array(img,dtype='float')
print(f"original size of image = {image_matrix.shape}")
```

```python
U, sigma , Vt = np.linalg.svd(image_matrix,full_matrices=False)
print(f"U shape: {U.shape}")
print(f"Sigma shape: {sigma.shape}")
print(f"Vt shape: {Vt.shape}")
```

```python
k1 = 30 # Number of singular values
k2 = 100
compressed_img1 = np.dot(U[:,:k1],np.dot(np.diag(sigma[:k1]),Vt[:k1,:]))
compressed_img2 = np.dot(U[:,:k2],np.dot(np.diag(sigma[:k2]),Vt[:k2,:]))
print(f"Compressed image matrix shape (approx): {compressed_img.shape}")
```
